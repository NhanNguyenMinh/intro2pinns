# TF-a · Mạng Keras là một hàm tọa độ → chuyển vị

Mục tiêu: hiểu shape, Dense, tanh, model(x) và trainable_variables. Chưa dùng phương trình dầm. Chạy trên CPU Colab.


```python
import tensorflow as tf
tf.keras.backend.set_floatx('float64')
tf.keras.utils.set_random_seed(7)
print('TensorFlow', tf.__version__)
```

## 1. Một lớp Dense

Một hàng x có một tọa độ. Dense(1,3) tạo ba đặc trưng: z=xW+b. In shape và tham số để nhìn thấy điều này.


```python
layer = tf.keras.layers.Dense(3)
x = tf.constant([[0.0], [0.5], [1.0]], dtype=tf.float64)
z = layer(x)
print('x:', x.shape, 'z:', z.shape)
print('kernel:', layer.kernel.shape, 'bias:', layer.bias.shape)
```

## 2. Ghép thành hàm u_theta(x)

Tanh tạo tính phi tuyến; Dense cuối không có activation. Đầu ra ở điểm mới là dự đoán của cùng một hàm, không phải một ô mới trong bảng nghiệm.


```python
model = tf.keras.Sequential([tf.keras.Input(shape=(1,)),
    tf.keras.layers.Dense(8, activation='tanh'),
    tf.keras.layers.Dense(1)])
y = model(x)
x_new = tf.constant([[0.37]], dtype=tf.float64)
print('y tại 3 điểm:', y.numpy().reshape(-1))
print('y tại 0.37:', model(x_new).numpy().reshape(-1))
print('số tham số:', sum(int(tf.size(v)) for v in model.trainable_variables))
```

Tự làm: đổi Dense(8) thành Dense(16), in lại số tham số và shape đầu ra. Trước train, không gọi các giá trị này là nghiệm dầm.
