# 01 · CMD và Python cơ bản

Mở notebook từ thư mục `learn-python`. Chạy các ô theo thứ tự. Sau đó chạy script cùng tên từ CMD và so sánh kết quả. Kiến thức tương ứng trong C: biến, hàm, `for`, `if`; Python không cần khai báo kiểu biến trước.

```python
length = 1.0
flexural_rigidity = 1.0
load = 1.0
points = [0.0, 0.25, 0.5, 0.75, 1.0]
parameters = {'L': length, 'EI': flexural_rigidity, 'q': load}
print(type(points), type(parameters), parameters)
```

Nghiệm chính xác của dầm tựa đơn chịu tải đều: $w(x)=qx(L^3-2Lx^2+x^3)/(24EI)$.

```python
def exact_deflection(x, L=1.0, EI=1.0, q=1.0):
    if not 0 <= x <= L:
        raise ValueError('x must be inside the beam')
    return q*x*(L**3 - 2*L*x**2 + x**3)/(24*EI)

for x in points:
    print(f'x={x:.2f}, w={exact_deflection(x):.8f}')
```

```python
assert abs(exact_deflection(0.0)) < 1e-12
assert abs(exact_deflection(1.0)) < 1e-12
assert abs(exact_deflection(0.5) - 5/384) < 1e-12
print('Three checks passed.')
```
