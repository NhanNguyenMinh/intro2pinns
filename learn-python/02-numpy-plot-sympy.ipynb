# 02 · NumPy, đồ thị, SymPy

NumPy xử lý cả lưới điểm, Matplotlib vẽ kết quả, SymPy kiểm tra biểu thức. Chạy sau bài 01.

```python
import numpy as np
a = np.array([[1., 2.], [3., 4.]])
b = np.array([[2., 0.], [0., 2.]])
print('elementwise:', a*b)
print('matrix product:', a@b)
```

```python
x = np.linspace(0.0, 1.0, 201)
w = (x**4 - 2*x**3 + x)/24
M = x*(1-x)/2
print('max w:', w.max(), 'max M:', M.max())
assert np.isclose(w[100], 5/384)
```

```python
import sympy as sp
s = sp.symbols('s', real=True)
w_symbolic = (s**4 - 2*s**3 + s)/24
assert sp.simplify(sp.diff(w_symbolic, s, 4)-1) == 0
assert sp.simplify(w_symbolic.subs(s,0)) == 0
assert sp.simplify(w_symbolic.subs(s,1)) == 0
assert sp.simplify(sp.diff(w_symbolic,s,2).subs(s,0)) == 0
assert sp.simplify(sp.diff(w_symbolic,s,2).subs(s,1)) == 0
print('Equation and four boundary conditions passed.')
```

```python
import matplotlib.pyplot as plt
from pathlib import Path
fig, axes = plt.subplots(1, 2, figsize=(9, 3))
axes[0].plot(x, w); axes[0].set(xlabel='x/L', ylabel='w', title='Deflection'); axes[0].grid(True)
axes[1].plot(x, M); axes[1].set(xlabel='x/L', ylabel='M', title='Moment'); axes[1].grid(True)
fig.tight_layout()
Path('outputs').mkdir(exist_ok=True)
fig.savefig('outputs/beam_reference_notebook.png', dpi=160)
plt.show()
```
